W3A13
[Tính tiền điện]
Nhập vào số kWh điện tiêu thụ. Tính tiền điện theo các bậc sau:
- 0-50 kWh: 1500đ/kWh
- 51-100 kWh: 2000đ/kWh
- 100 kWh: 3000đ/kWh 

```python
n = int(input())
if n <= 50:
    cost = n * 1500
elif n <= 100:
    cost = 50 * 1500 + (n - 50) * 2000
else:
    cost = 50 * 1500 + 50 * 2000 + (n - 100) * 3000
print("Tiền điện phải trả:", cost, "đ")
```

W3A16
[Rounding]
Nhập vào một số thực: in ra kết quả làm tròn lên ,tròn xuống, làm tròn tới số nguyên gần nhất (không được sử dụng hàm
sẵn có của pyton). Các kết quả cách nhau một dấu cách 

```python
n = float(input())
if n >=0:
    lam_tron = n//1
    tron_xuong = n//1
    if n -int(n) >= 0.5:
        tron_len = int(n) + 1
        print(lam_tron, tron_xuong, tron_len)
    else:
        tron_len = int(n)
        print(lam_tron, tron_xuong, tron_len)
else:
    if n - int(n)<=0.5:
        print(int(n), int(n)-1, int(n))
    else:
        print(int(n), int(n)-1, int(n))
```

W3A17
Cấp số nhân
Cho 4 số a, b, c, d. Hãy kiểm tra xem 4 số này có thể theo thứ tự tạo thành 1 cấp số nhân với công bội nguyên theo đúng
thứ tự a, b, c, d hay không?
Gợi ý : Tìm công bội (b / a) nếu b chia hết cho a, sau đó lấy b nhân công bội và so sánh vs c, c nhân công bội và so sánh vs
d.

```python
a,b,c,d = map(float, input().split())
if a != 0 and b % a == 0:  
    q = b // a
    if b * q == c and c * q == d:
        print("Cấp số nhân với công bội:", q)
    else:
        print("Không phải cấp số nhân")
else:
    print("Không phải cấp số nhân")
```
